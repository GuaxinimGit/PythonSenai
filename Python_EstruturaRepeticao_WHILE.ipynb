```python
# DESAFIO 02
"""
Faça um programa onde o computador vai “pensar” em um número entre 0 a 10.
O jogador vai tentar adivinhar até acertar, mostrando no final quantos
palpites foram necessários para vencer.
"""
import random
from google.colab import output

computador = random.randint(0,10)
tentativas = 0

while True:
    resposta = int(input("Digite um número entre 0 e 10: "))
    tentativas += 1
    if resposta != computador:
        print("\033[31mResposta incorreta! Tente novamente\033[0m")
        continue
    if resposta == computador:
        output.clear()
        print(f"\033[32mParabéns, você acertou!\n"
              f"\033[33mNúmero correto: {resposta} | Quantidade de tentativas: {tentativas}")
        break
```

## Desafio 03

```python
# DESAFIO 03
"""
Crie um programa que leia dois valores e mostre um menu na tela:
[1] Somar
[2] Multiplicar
[3] Maior
[4] Novos Números
[5] Sair do programa

Seu programa deverá realizar a operação solicitada em cada caso.
"""

while True:
    valor1 = float(input("Digite o 1° valor: "))
    valor2 = float(input("Digite o 2° valor: "))
    menu = int(input("""
[1] Somar
[2] Multiplicar
[3] Maior
[4] Novos Números
[5] Sair do programa

"""))
    if menu == 1:
        print(f"\nA soma dos valores é: {valor1+valor2}")
        break
    if menu == 2:
        print(f"\nA multiplicação dos valores é: {valor1*valor2}")
        break
    if menu == 3:
        if valor1 > valor2:
            print(f"\nO maior valor é: {valor1}")
        else:
            print(f"\nO maior valor é: {valor2}")
        break
    if menu == 4:
        print("\nInsira novos valores...\n")
        continue
    if menu == 5:
        print("\nPrograma finalizado... ")
        break
    if menu > 5 or menu == 0:
        print("\nOpção inválida!\n")
        continue
```

```python
# DESAFIO 03 - V2

while True:
    valor1 = float(input("Digite o 1° valor: "))
    valor2 = float(input("Digite o 2° valor: "))
    menu = int(input("""
[1] Somar
[2] Multiplicar
[3] Maior
[4] Novos Números
[5] Sair do programa

"""))
    if menu == 1:
        print(f"\nA soma dos valores é: {valor1+valor2}")
        break
    if menu == 2:
        print(f"\nA multiplicação dos valores é: {valor1*valor2}")
        break
    if menu == 3:
        print(f"\nO maior valor: {max(valor1, valor2)}")
        break
    if menu == 4:
        print("\nInsira novos valores...\n")
        continue
    if menu == 5:
        print("\nPrograma finalizado...")
        break
    if menu > 5 or menu == 0:
        print("\n\033[31mOpção inválida! Tente novamente...\033[0m\n")
        continue
```

```python
# DESAFIO 03 - V3

from google.colab import output

while True:
    print()
    num1 = float(input("Digite o 1° número: "))
    num2 = float(input("Digite o 2° número: "))
    oper = input("""
[1] Somar
[2] Multiplicar
[3] Maior
[4] Novos Números
[5] Sair do programa

""")
    match oper:
        case '1':
            output.clear()
            print(f"\n{num1} + {num2} = {num1+num2}")
        case '2':
            output.clear()
            print(f"\n{num1} * {num2} = {num1*num2}")
        case '3':
            output.clear()
            print(f"\nO maior número entre {num1} e {num2} = {max(num1,num2)}")
        case '4':
            output.clear()
            print("\nInsira novos valores...")
            continue
        case '5':
            output.clear()
            print("\nPrograma finalizado!")
            break
        case _:
            output.clear()
            print("\nOpção inválida! Tente novamente...")
```

## Desafio 04

```python
# DESAFIO 04
"""
Faça um programa que leia um número qualquer e mostre o seu
fatorial.
Exemplo:

5! = 5 x 4 x 3 x 2 x 1 = 120
"""

fat = 1
num = int(input("Digite um número: "))
while num > 0:
    fat *= num
    num -= 1
    print(fat)
    if num == 1:
        break
```

```python
# DESAFIO 04 - V2

from google.colab import output

fat = 1
num = int(input("Digite um número: "))
output.clear()
print(f"O fatorial de {num}! é: \n{num}", end='')
while num > 0:
    fat *= num
    num -= 1
    print(f" X {num}", end='')
    if num == 1:
        print(f" = \033[33m{fat}")
        break
```

```python
# DESAFIO 04 - V3

fatorial = 1
numero = int(input("Dígite um número para ver seu Fatorial: "))
saida = str(numero) + "! = " # concatena o texto.
while numero >= 1:
    if numero == 1:
        saida = saida + str(numero) + ' ='
    else:
        saida = saida + str(numero) + ' x '
    fatorial *= numero
    numero -= 1
print(f"\n\033[1;34m{saida} {fatorial:,.0f}")
```

```python
# DESAFIO 04 - V4

fatorial = 1

num = int(input("Digite um número para ver seu Fatorial: "))
r = f'{num}! = '
while num >= 1:
    fatorial*=num
    if num > 1:
        r += f'{num} x '
    else:
        r += f'1 = {fatorial}'
    num-=1
print(f'\n{r}')
```

```python
# DESAFIO 04 - V5

tentativas = False
fat = 1
num = int(input("Digite um número: "))
txt = f"{num}! = "

while num > 0:
    if tentativas:
        txt += "x "
    txt += f"{num} "
    fat *= num
    num -= 1
    tentativas = True
print(f"{txt}= \033[1m{fat:,}\033[0m")
```

## Desafio 08

```python
# DESAFIO 08
"""
Crie um programa que leia vários números inteiros pelo teclado. No
final da execução, mostre a média entre todos os valores e qual foi o
maior e o menor valores lidos. O programa deve perguntar ao usuário
se ele quer ou não continuar a digitar valores
"""

tentativas = 0
soma = 0
max = 0
min = float('inf')

while True:
    num = int(input("Digite um número inteiro: "))
    soma += num
    tentativas += 1
    media = soma/tentativas
    if num > max:
        max = num
    if num < min:
        min = num
    resposta = input("Deseja digitar outro número ou sair (s/n)? ").lower()
    if resposta == 's':
        continue
    elif resposta == 'n':
        print(f"\nO {max} foi maior número, \nO {min} foi o menor número \n{media:,.1f} é a média entre eles.")
        break
    else:
        print("Opção inválida!")
        continue
```

```python
# DESAFIO 08 - V2

soma = 0
qtd = 0
maior = 0
menor = float('inf')
while True:
    num = int(input("Digite um número inteiro: "))
    qtd+=1
    soma+=num
    if num > maior:
        maior = num
    if num < menor:
        menor = num
    op = input(f"Deseja continuar? (S/N) ").upper()
    if op == 'S':
        continue
    elif op == 'N':
        print("\nFinalizando...")
        break
    else:
        print("Opção inválida...")
        break
print(f"\nMaior: \033[0;31m{maior}\033[0m"
      f"\nMenor: \033[0;31m{menor}\033[0m"
      f"\nMédia: \033[0;31m{soma/qtd:,.1f}\033[0m")
```

```python
# DESAFIO 08 - V3

soma  = 0
qtd   = 0
maior = float('-inf')
menor = float('inf')
while True:
    num = int(input("Digite um número inteiro (999 para sair): "))
    if num == 999:
        print("\nSaindo...")
        break
    else:
        maior = max(maior, num)
        menor = min(menor, num)
        soma+=num
        qtd+=1
print(f"\n\033[1;33mMédia: {soma/qtd:,.1f}" if qtd>0 else "Média: 0")
print(f"\033[1;32mMaior: {maior}" if qtd>0 else "Maior: 0")
print(f"\033[1;31mMenor: {menor}" if qtd>0 else "Menor: 0")
```

## Desafio 09

```python
# DESAFIO 09
"""
Escreva um programa que leia um número n inteiro qualquer e mostre
na tela os n primeiros elementos de uma sequência de Fibonacci.
Exemplo:

0 –> 1 –> 1 –> 2 –> 3 –> 5 –> 8
1+1 = 2
1+2 = 3
2+3 = 5
"""
anterior = 0
atual = 1
fib = '0  1'
cont = 0

n = int(input("Digite um número: "))
while cont < n-2:
    numero = anterior + atual
    fib = fib + '  ' +f"{numero:,.0f}"
    anterior = atual
    atual = numero
    cont += 1
print(f"\n\033[1;36m{fib}")
```

```python
# DESAFIO 09 - V2

anterior = 0
atual = 1
fibo = 0
i = 0
num = int(input("Digite uma quantidade para sequência Fibonacci: "))
while i < num:
    print(fibo, end=' ')
    fibo=atual+anterior
    anterior=atual
    atual=fibo
    i+=1
```

```python
# DESAFIO 09 - V3

f1 = 0
f2 = 1
fib = 0
texto = f"{f1} -> {f2}"
numero = int(input("Digite um número: "))

i = 0
while i < numero - 2:
    fib = f1 + f2
    texto += f" -> {fib}"
    f1 = f2
    f2 = fib
    i += 1
print(f"\n\033[1m{texto}\033[0m")
```
